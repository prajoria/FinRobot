# FinRobot Practical Examples

This notebook demonstrates how to use FinRobot for various financial analysis tasks.

## Prerequisites
1. Configure `OAI_CONFIG_LIST` with your OpenAI API key
2. Configure `config_api_keys` with your financial data API keys
3. Install FinRobot: `pip install -e .`

```python
# Import required libraries
import autogen
import os
from finrobot.utils import get_current_date, register_keys_from_json
from finrobot.agents.workflow import SingleAssistant, SingleAssistantShadow

print("✅ Imports successful!")
print(f"📅 Current date: {get_current_date()}")
```

```python
# Setup configuration
llm_config = {
    "config_list": autogen.config_list_from_json(
        "../OAI_CONFIG_LIST",
        filter_dict={"model": ["gpt-4-0125-preview"]},
    ),
    "timeout": 120,
    "temperature": 0,
}

# Register API keys
register_keys_from_json("../config_api_keys")
print("✅ Configuration complete!")
```

## Example 1: Simple Stock Analysis

```python
# Create Market Analyst agent
analyst = SingleAssistant(
    "Market_Analyst",
    llm_config,
    human_input_mode="NEVER",
)

# Choose stock to analyze
stock = "AAPL"  # Change this to any stock you want to analyze

# Analysis prompt
prompt = f"""
Analyze {stock} stock for {get_current_date()}.

Please provide:
1. Current price and recent performance (1 week, 1 month)
2. Key financial metrics (P/E ratio, market cap, etc.)
3. Recent news and market sentiment
4. Technical analysis (support/resistance levels)
5. Short-term prediction (next 2 weeks) with reasoning
6. Risk factors to consider

Keep the analysis concise but comprehensive.
"""

print(f"🔍 Analyzing {stock}...")
result = analyst.chat(prompt)
```

## Example 2: Sector Comparison Analysis

```python
# Compare multiple stocks in the same sector
tech_stocks = ["AAPL", "MSFT", "GOOGL", "NVDA"]

comparison_prompt = f"""
Compare these technology stocks: {', '.join(tech_stocks)}

For each stock, provide:
1. Current valuation metrics (P/E, P/S ratios)
2. Recent financial performance
3. Growth prospects and catalysts
4. Risk assessment

Then rank them by:
- Best value opportunity
- Highest growth potential
- Lowest risk profile
- Overall investment attractiveness

Provide specific recommendations for different investor profiles.
"""

print("📊 Comparing technology stocks...")
comparison = analyst.chat(comparison_prompt)
```

## Example 3: Portfolio Health Check

```python
# Sample portfolio for analysis
portfolio = {
    "AAPL": 25,   # 25% allocation
    "MSFT": 20,   # 20% allocation
    "GOOGL": 15,  # 15% allocation
    "AMZN": 15,   # 15% allocation
    "TSLA": 10,   # 10% allocation
    "NVDA": 10,   # 10% allocation
    "SPY": 5,     # 5% allocation (ETF)
}

portfolio_prompt = f"""
Analyze this portfolio composition:
{chr(10).join([f'- {stock}: {allocation}%' for stock, allocation in portfolio.items()])}

Provide comprehensive analysis:

1. DIVERSIFICATION ANALYSIS:
   - Sector concentration risk
   - Geographic exposure
   - Market cap distribution

2. RISK ASSESSMENT:
   - Portfolio volatility estimate
   - Correlation between holdings
   - Concentration risk (top 3 holdings)

3. PERFORMANCE METRICS:
   - Recent performance vs S&P 500
   - Risk-adjusted returns
   - Dividend yield of portfolio

4. REBALANCING RECOMMENDATIONS:
   - Overweight/underweight positions
   - Suggested adjustments
   - New positions to consider

5. MARKET OUTLOOK IMPACT:
   - How current market conditions affect this portfolio
   - Upcoming catalysts or risks
   - Defensive vs growth positioning
"""

print("💼 Analyzing portfolio...")
portfolio_analysis = analyst.chat(portfolio_prompt)
```

## Example 4: Market Sentiment Analysis

```python
# Analyze market sentiment for different themes
sentiment_prompt = """
Analyze current market sentiment and trends for these investment themes:

1. ARTIFICIAL INTELLIGENCE:
   - Key stocks: NVDA, AMD, GOOGL, MSFT
   - Recent developments and news sentiment
   - Market positioning and valuations

2. ELECTRIC VEHICLES:
   - Key stocks: TSLA, F, GM, RIVN
   - Industry trends and adoption rates
   - Competitive landscape changes

3. CLOUD COMPUTING:
   - Key stocks: AMZN, MSFT, CRM, SNOW
   - Enterprise adoption trends
   - Competitive positioning

4. RENEWABLE ENERGY:
   - Key stocks: ENPH, SEDG, NEE, BEP
   - Policy impacts and government support
   - Technology advancement trends

For each theme, provide:
- Current sentiment score (Bullish/Neutral/Bearish)
- Key catalysts and risks
- Best positioned companies
- Investment timeline recommendations

Conclude with the most attractive theme for the next 6 months.
"""

print("🌊 Analyzing market sentiment...")
sentiment_analysis = analyst.chat(sentiment_prompt)
```

## Example 5: Options Strategy Analysis

```python
# Analyze options strategies for a specific stock
options_stock = "AAPL"

options_prompt = f"""
Analyze options strategies for {options_stock} based on current market conditions:

1. CURRENT SETUP:
   - Current stock price and recent volatility
   - Implied volatility levels vs historical
   - Upcoming events (earnings, product launches, etc.)

2. STRATEGY RECOMMENDATIONS:
   
   For BULLISH outlook:
   - Call options strategy
   - Bull call spreads
   - Covered calls (if holding stock)
   
   For NEUTRAL outlook:
   - Iron condor strategy
   - Butterfly spreads
   - Theta decay strategies
   
   For BEARISH outlook:
   - Put options strategy
   - Bear put spreads
   - Protective puts

3. RISK MANAGEMENT:
   - Position sizing recommendations
   - Stop-loss levels
   - Profit-taking targets
   - Time decay considerations

4. TIMING ANALYSIS:
   - Best expiration dates
   - Strike price selection
   - Entry and exit timing

Provide specific strike prices and expiration dates for each strategy.
"""

print(f"📈 Analyzing options strategies for {options_stock}...")
options_analysis = analyst.chat(options_prompt)
```

## Example 6: Economic Calendar Analysis

```python
# Analyze upcoming economic events and their market impact
economic_prompt = f"""
Analyze the economic calendar for the next 2 weeks from {get_current_date()}:

1. KEY ECONOMIC EVENTS:
   - Federal Reserve meetings/announcements
   - Employment reports (Non-farm payrolls, unemployment)
   - Inflation data (CPI, PPI)
   - GDP releases
   - Consumer sentiment indicators

2. EARNINGS CALENDAR:
   - Major companies reporting earnings
   - Sectors with heavy concentration of reports
   - Key metrics to watch for each sector

3. MARKET IMPACT ANALYSIS:
   - Which events are likely to cause volatility
   - Sector-specific impacts
   - Currency and bond market implications

4. TRADING STRATEGIES:
   - Positions to avoid during high-impact events
   - Volatility plays around key announcements
   - Safe-haven assets to consider

5. PREPARATION CHECKLIST:
   - Portfolio adjustments to make before events
   - Hedging strategies to consider
   - Opportunities to watch for

Focus on actionable insights for active traders and long-term investors.
"""

print("📅 Analyzing economic calendar...")
economic_analysis = analyst.chat(economic_prompt)
```
